```python
import requests
import xml.etree.ElementTree as ET

def getWBdata(countrycode, indicator):
    # Construct the URL for the World Bank API
    url = f"https://api.worldbank.org/v2/countries/{countrycode}/indicator/{indicator}?format=xml"
    
    # Fetch the data from the API
    response = requests.get(url)
    if response.status_code != 200:
        print("Error fetching data from the World Bank API.")
        return
    
    # Parse the XML data
    root = ET.fromstring(response.content)
    
    # Extract the indicator text
    indicator_element = root.find(".//{http://www.worldbank.org}indicator")
    if indicator_element is not None:
        indicator_text = indicator_element.text
        print(f"Indikaattori: {indicator_text}")
    else:
        print("Indicator not found.")
        return
    
    # Extract the country name
    country_element = root.find(".//{http://www.worldbank.org}country")
    if country_element is not None:
        country_name = country_element.text
        print(f"Maa: {country_name}")
    else:
        print("Country not found.")
        return
    
    # Extract year and value pairs
    data = []
    for value_element in root.findall(".//{http://www.worldbank.org}value"):
        year = value_element.attrib.get("date")
        value = value_element.text
        if value:  # Skip empty values
            data.append((int(year), float(value)))
    
    # Sort the data by year and print it
    for year, value in sorted(data):
        print(f"{year} {value}")
```

```python
import requests
import xml.etree.ElementTree as ET

def getWBdata(countrycode, indicator):
    # Base URL for the World Bank API
    base_url = f"https://api.worldbank.org/v2/countries/{countrycode}/indicator/{indicator}?format=xml"
    
    # Initialize variables for pagination
    page = 1
    total_pages = 1
    namespace = {"wb": "http://www.worldbank.org"}
    data = []

    while page <= total_pages:
        # Fetch the data for the current page
        url = f"{base_url}&page={page}"
        response = requests.get(url)
        if response.status_code != 200:
            print("Error fetching data from the World Bank API.")
            return
        
        # Parse the XML data
        root = ET.fromstring(response.content)
        
        # Get the total number of pages (only on the first request)
        if page == 1:
            total_pages = int(root.attrib.get("pages", 1))
        
        # Extract year and value pairs
        for data_element in root.findall(".//wb:data", namespace):
            year = data_element.find("wb:date", namespace).text
            value_element = data_element.find("wb:value", namespace)
            if value_element is not None and value_element.text:  # Skip empty values
                value = float(value_element.text)
                data.append((int(year), value))
        
        # Move to the next page
        page += 1

    # Extract the indicator text
    indicator_element = root.find(".//wb:indicator", namespace)
    if indicator_element is not None:
        indicator_text = indicator_element.text
        print(f"Indikaattori: {indicator_text}")

    # Extract the country name
    country_element = root.find(".//wb:country", namespace)
    if country_element is not None:
        country_name = country_element.text
        print(f"Maa: {country_name}")

    # Sort the data by year and print it
    for year, value in sorted(data):
        print(f"{year} {value}")
```

```python
import xml.etree.ElementTree as ET
from urllib.request import urlopen

def getWBdata(countrycode, indicator):
    base_url = f"https://api.worldbank.org/v2/countries/{countrycode}/indicator/{indicator}?format=xml"
    
    page = 1
    total_pages = 1
    namespace = {"wb": "http://www.worldbank.org"}
    data = []

    while page <= total_pages:
        # Construct the URL for the current page
        url = f"{base_url}&page={page}"
        
        # Fetch the data using urllib
        with urlopen(url) as response:
            content = response.read()
        
        # Parse the XML data
        root = ET.fromstring(content)
        
        # Get the total number of pages (only on the first request)
        if page == 1:
            total_pages = int(root.attrib.get("pages", 1))
        
        # Extract year and value pairs
        for data_element in root.findall(".//wb:data", namespace):
            year = data_element.find("wb:date", namespace).text
            value_element = data_element.find("wb:value", namespace)
            if value_element is not None and value_element.text:  # Skip empty values
                value = float(value_element.text)
                data.append((int(year), value))
        
        # Move to the next page
        page += 1

    # Extract the indicator text
    indicator_element = root.find(".//wb:indicator", namespace)
    if indicator_element is not None:
        indicator_text = indicator_element.text
        print(f"Indikaattori: {indicator_text}")

    # Extract the country name
    country_element = root.find(".//wb:country", namespace)
    if country_element is not None:
        country_name = country_element.text
        print(f"Maa: {country_name}")

    # Sort the data by year and print it
    for year, value in sorted(data):
        print(f"{year} {value:.15g}")
```
