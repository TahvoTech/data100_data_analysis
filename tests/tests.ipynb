```python
def run_quiz(statements_by_topic):
    print("Welcome to the Quiz!")
    print("Available topics:")
    for topic in statements_by_topic.keys():
        print(f"- {topic}")
    
    # Let the user choose a topic
    chosen_topic = input("Choose a topic: ").strip()
    if chosen_topic not in statements_by_topic:
        print("Invalid topic. Exiting.")
        return
    
    print(f"\nYou chose: {chosen_topic}")
    print("Answer True or False for each statement.\n")
    
    # Get the statements for the chosen topic
    statements = statements_by_topic[chosen_topic]
    
    # Loop through the statements
    for i, item in enumerate(statements, start=1):
        statement = item["statement"]
        correct_answer = item["answer"]
        explanation = item["explanation"]
        
        print(f"Statement {i}: {statement}")
        user_answer = input("Your answer (True/False): ").strip()
        
        if user_answer.lower() == correct_answer.lower():
            print("Correct!")
        else:
            print("Incorrect.")
        print(f"Explanation: {explanation}\n")

# Load the statements by topic

with open("statements_by_topic.json", "r", encoding="utf-8") as f:

    # Run the quiz
    run_quiz(statements_by_topic)
```
