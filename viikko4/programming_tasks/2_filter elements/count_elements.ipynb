Question

Toteuta Python-funktio activePostOffices(filename, encoding="utf-8"), joka avaa parametrin filename nimeämän XML-tiedoston käyttäen parametrin encoding mukaista merkistöä. Funktio olettaa, että tiedosto sisältää samanmuotoista XML-dataa postitoimipaikoista, mitä kurssilla on käsitelty. Funktio palauttaa listan niistä postitoimipaikka-elementeistä, joita ei ole lakkautettu. Postitoimipaikkaa ei ole lakkautettu, jos sitä vastaavan postitoimipaikka-elementin sisältämä lakkauttamispäivämäärä-elementti on tyhjä.

Tehtävä tulee ratkaista soveltamalla funktiota filter XML-datan postitoimipaikka-elementtejä läpikäyvään iter-kutsuun. Koska tarkoitus on käyttää funktiota filter, toteutuksesi ei saa sisältää yhtään for- tai while-silmukkaa (eikä myöskään ns. "list comprehension"-lauseketta).

Toteuta ratkaisusi täydentämällä valmiiseen koodipohjan syötetiedoston avaaminen XML-datan läpikäymiseksi sekä filter-kutsun sisältö.

```python
import xml.etree.ElementTree as ET

def activePostOffices(filename, encoding="utf-8"):
    with open(filename, encoding=encoding) as file:
        data = file.read()
        root = ET.fromstring(data)
        iteration = root.iter("toimipaikka")
    return list(filter(lambda x: x.find("lakkauttamispäivämäärä").text is None, iteration))
```

```python
import xml.etree.ElementTree as ET

def XMLelementCount(filename, element=None, encoding='utf-8'):
    #Toteuta funktion runko
    with open(filename, encoding=encoding) as file:
        tree = ET.parse(file)
        root = tree.getroot()
        
        if element is None:
            count = sum(1 for _ in root.iter())
            print(f"Tiedostossa '{filename}' on {count} elementtiä")
        else:
            count = sum(1 for _ in root.iter(element))
            print(f"Tiedostossa '{filename}' on {count} {element}-elementtiä")
```
