Kirjoita Python-ohjelma, joka saa syötteenään kolme XML-syötetiedostoa, jotka sisältävät tietoa joidenkin alueiden pinta-aloista, asukasluvuista ja bruttokansantuotteista. Ohjelma luo tietojen pohjalta uuden XML-tiedoston, joka esittää alueiden tiedot yhteenkootusti seuraavanlaisessa muodossa:

```xml
<regions>
  <region>
    <name>Guam</name>
    <area>540</area>
    <population>168783</population>
    <gdp>5259729936.30573</gdp>
  </region>
  <!--- Edellisen elementin perässä vielä lisää region-elementtejä, jos alueita enemmän kuin yksi. -->
</regions>
```


Eli XML-tiedoston juurielementin nimi on "regions", ja juurielementin alla on yksi region-elementti kutakin aluetta kohti. Kunkin region-elementin sisällä on alueen nimen, pinta-alan, asukasluvun ja bruttokansantuotteen ilmaisevat name-, area-, population- ja gdp-elementit. Elementit esitetään alueiden nimien mukaan kasvavassa aakkosjärjestyksessä. Tiedostoon ei tule sisällyttää edellä toiseksi viimeisellä rivillä olevaa XML-kommenttiriviä, sen on vain tarkoitus tarkentaa tietojen esitysmuotoa. Kutsu funktiota ET.indent() ennen XML-datan kirjoittamista tiedostoon, jotta tulostiedoston elementit tulevat ihmisten näkökulmasta luettavampaan "nätisti rivitettyyn / sisennettyyn" muotoon.

Syötetiedostot ovat rakenteeltaan samankaltaisia kuin Moodlessa viikon 2 alaisuudessa olevat esimerkkitiedostot area1.xml, population1.xml ja gdp1.xml. Ohjelmasi saa syötetiedostojen sekä luotavan uuden XML-tulostiedoston nimet komentoriviparametreina koodipohjassa esitetyllä tavalla.


````python
from sys import argv
import xml.etree.ElementTree as ET

if __name__ == "__main__":
    areafn = argv[1] # Pinta-alat sisältävän XML-tiedoston nimi
    popfn = argv[2]  # Asukasluvut sisältävän XML-tiedoston nimi
    gdpfn = argv[3]  # Bruttokansantuotteet sisältävän XML-tiedoston nimi
    resultfn = argv[4] # Tulostiedoston nimi
    # Täydennä vastauksesi tähän.
````

````python
from sys import argv
import xml.etree.ElementTree as ET

if __name__ == "__main__":
    areafn = argv[1] # Pinta-alat sisältävän XML-tiedoston nimi
    popfn = argv[2]  # Asukasluvut sisältävän XML-tiedoston nimi
    gdpfn = argv[3]  # Bruttokansantuotteet sisältävän XML-tiedoston nimi
    resultfn = argv[4] # Tulostiedoston nimi
    # Täydennä vastauksesi tähän.

    # parse files
    area_tree = ET.parse(areafn)
    area_root = area_tree.getroot()

    pop_tree = ET.parse(popfn)
    pop_root = pop_tree.getroot()

    gdp_tree = ET.parse(gdpfn)
    gdp_root = gdp_tree.getroot()

    # make dictionaries for each data
    area_dict = {}
    pop_dict = {}
    gdp_dict = {}

    # extract area data to area_dict
    for record in area_root.find("data"):
        country = None
        area = None
        for field in record:
            if field.attrib.get("name") == "Country or Area":
                country = field.text
            elif field.attrib.get("name") == "Value":
                area = str(field.text)
        if country and area:
            area_dict[country] = area

    # extract population data to pop_dict
    for record in pop_root.find("data"):
        country = None
        population = None
        for field in record:
            if field.attrib.get("name") == "Country or Area":
                country = field.text
            elif field.attrib.get("name") == "Value":
                population = int(field.text)
        if country and population:
            pop_dict[country] = population

    # extract gdp data to gdp_dict
    for record in gdp_root.find("data"):
        country = None
        gdp = None
        for field in record:
            if field.attrib.get("name") == "Country or Area":
                country = field.text
            elif field.attrib.get("name") == "Value":
                gdp = float(field.text)
        if country and gdp:
            gdp_dict[country] = gdp

    # create new xlm data
    regions_root = ET.Element("regions")

    # sort countries by name
    sorted_countries = sorted(area_dict.keys())

    # add data to new xml
    for country in sorted_countries:
        if country in pop_dict and country in gdp_dict:
            region = ET.SubElement(regions_root, "region")
            ET.SubElement(region, "name").text = country
            ET.SubElement(region, "area").text = str(area_dict[country])
            ET.SubElement(region, "population").text = str(pop_dict[country])
            ET.SubElement(region, "gdp").text = f"{gdp_dict[country]:.5f}"

    # use indent -> pretty print
    ET.indent(regions_root)

    # write file
    tree = ET.ElementTree(regions_root)
    tree.write(resultfn, encoding="utf-8")
````

````python
import xml.etree.ElementTree as ET

if __name__ == "__main__":
    areafn = "area1.xml"
    popfn = "population1.xml"
    gdpfn = "gdp1.xml"
    resultfn = "new_file.xml"  # Output file name

    # Parse the input XML files
    area_tree = ET.parse(areafn)
    area_root = area_tree.getroot()

    pop_tree = ET.parse(popfn)
    pop_root = pop_tree.getroot()

    gdp_tree = ET.parse(gdpfn)
    gdp_root = gdp_tree.getroot()

    # Create dictionaries to store the data
    area_dict = {}
    pop_dict = {}
    gdp_dict = {}

    # Extract area data
    for record in area_root.find("data"):
        country = None
        area = None
        for field in record:
            if field.attrib.get("name") == "Country or Area":
                country = field.text
            elif field.attrib.get("name") == "Value":
                area = int(field.text)
        if country and area:
            area_dict[country] = area

    # Extract population data
    for record in pop_root.find("data"):
        country = None
        population = None
        for field in record:
            if field.attrib.get("name") == "Country or Area":
                country = field.text
            elif field.attrib.get("name") == "Value":
                population = int(field.text)
        if country and population:
            pop_dict[country] = population

    # Extract GDP data
    for record in gdp_root.find("data"):
        country = None
        gdp = None
        for field in record:
            if field.attrib.get("name") == "Country or Area":
                country = field.text
            elif field.attrib.get("name") == "Value":
                gdp = float(field.text)
        if country and gdp:
            gdp_dict[country] = gdp

    # Create the root element for the new XML file
    regions_root = ET.Element("regions")

    # Combine the data and create region elements
    for country in sorted(area_dict.keys()):
        if country in pop_dict and country in gdp_dict:
            region = ET.SubElement(regions_root, "region")
            ET.SubElement(region, "name").text = country
            ET.SubElement(region, "area").text = str(area_dict[country])
            ET.SubElement(region, "population").text = str(pop_dict[country])
            ET.SubElement(region, "gdp").text = f"{gdp_dict[country]:.5f}"

    # Indent the XML for better readability
    ET.indent(regions_root)

    # Write the new XML data to the output file
    tree = ET.ElementTree(regions_root)
    tree.write(resultfn, encoding="utf-8", xml_declaration=True)

    print(f"New XML file '{resultfn}' created successfully.")
````

    - The extracted data is stored in dictionaries (`area_dict`, `pop_dict`, and `gdp_dict`) with the country name as the key.

### Why Use `<data>` and `<regions>`?

#### 1. **`<data>` in Input Files**:
- The `<data>` element in the input files (`area1.xml`, `population1.xml`, and `gdp1.xml`) serves as a container for all `<record>` elements.
- Each `<record>` represents a single region and contains multiple `<field>` elements with attributes like `name="Country or Area"` and `name="Value"`.
- The `<data>` element organizes the records and makes it easier to iterate over them in the program.

#### Example from `area1.xml`:


````python
<Root xmlns:wb="http://www.worldbank.org">
  <data>
    <record>
      <field name="Country or Area">Guam</field>
      <field name="Value">540</field>
    </record>
    <record>
      <field name="Country or Area">Guyana</field>
      <field name="Value">214970</field>
    </record>
  </data>
</Root>
````



#### 2. **`<regions>` in the Output File**:
- The `<regions>` element in the output file (`new_file.xml`) serves as the root element for the combined data.
- Each `<region>` element represents a single region and contains child elements for the region's name, area, population, and GDP.
- The `<regions>` element organizes the data hierarchically, making it easier to read and process.
