Toteuta Python-ohjelma, joka saa syötteenään kaksi XML-tiedostoa, jotka sisältävät tietoa joidenkin alueiden pinta-aloista ja asukasluvuista. Ohjelman tulee tulostaa ruudulle kunkin alueen asukastiheys (= asukasluku jaettuna pinta-alalla). Kutakin syötedatassa täysin edustettua (löytyy kumpikin tieto) aluetta kohden tulostetaan muotoa "alue asukastiheys" oleva rivi, missä asukastiheys on esitetty yhden desimaalin tarkkuudella. Rivit tulostetaan alueiden nimien mukaan kasvavassa aakkosjärjestyksessä.

Ohjelma saa syötetiedostojen nimet komentoriviparametreina (katso koodipohjaa). Syötetiedostojen oletetaan olevan samaa muotoa kuin viikon 2 alaisuudessa ladattavissa olevat tiedostot "area1.xml" ja "population1.xml". Tiedostot käyttävät UTF-8-merkistöä.


```python
from sys import argv
import xml.etree.ElementTree as ET

if __name__ == "__main__":
    areafn = argv[1] # Pinta-alat sisältävän XML-tiedoston nimi
    popfn = argv[2]  # Asukasluvut sisältävän XML-tiedoston nimi
    # Täydennä vastauksesi tähän.
```

```python
from sys import argv
import xml.etree.ElementTree as ET

if __name__ == "__main__":
    # areafn = argv[1]  # Pinta-alat sisältävän XML-tiedoston nimi
    # popfn = argv[2]   # Asukasluvut sisältävän XML-tiedoston nimi

    # Directly assign the filenames
    areafn = "area1.xml"  # Replace with the actual filename of the area XML file
    popfn = "population1.xml"  # Replace with the actual filename of the population XML file


    # Parse the area XML file
    area_tree = ET.parse(areafn)
    area_root = area_tree.getroot()

    # Parse the population XML file
    pop_tree = ET.parse(popfn)
    pop_root = pop_tree.getroot()

    # Create dictionaries to store area and population data
    area_dict = {}
    pop_dict = {}

    # Extract area data
    for area in area_root:
        name = area.find("Country or Area").text
        size = float(area.find("Value").text)
        area_dict[name] = size

    # Extract population data
    for population in pop_root:
        name = population.find("name").text
        count = int(population.find("count").text)
        pop_dict[name] = count

    # Calculate population density for regions with both area and population data
    densities = {}
    for name in area_dict:
        if name in pop_dict:
            density = pop_dict[name] / area_dict[name]
            densities[name] = density

    # Sort regions by name and print the results
    for name in sorted(densities.keys()):
        print(f"{name} {densities[name]:.1f}")
```

This code snippet is a Python script designed to calculate and display the population density for various regions based on data from two XML files: one containing area information (`area1.xml`) and the other containing population information (`population1.xml`). The script is executed when the file is run directly, as indicated by the `if __name__ == "__main__":` block.

### File Parsing
The script begins by assigning the filenames of the XML files to the variables `areafn` and `popfn`. These files are parsed using the `ET.parse()` method from the `xml.etree.ElementTree` module, which creates tree structures for the XML data. The root elements of these trees are stored in `area_root` and `pop_root`, respectively.

### Data Extraction
Two dictionaries, `area_dict` and `pop_dict`, are initialized to store the area and population data for each region. The script iterates over the `<data>` elements in both XML files. For each `<record>` in the area file, it extracts the country name (from the field with the attribute `name="Country or Area"`) and the area size (from the field with the attribute `name="Value"`). The extracted data is stored in `area_dict` with the country name as the key and the area size as the value. A similar process is followed for the population file, where the population count is extracted and stored in `pop_dict`.

### Population Density Calculation
The script calculates the population density for regions that have both area and population data. This is done by dividing the population value by the area size for each country present in both dictionaries. The results are stored in a new dictionary, `densities`, with the country name as the key and the calculated density as the value.

### Output
Finally, the script sorts the regions alphabetically by name and prints the population density for each region. The output is formatted to display the country name followed by the density value rounded to one decimal place.

### Use Case
This script is useful for analyzing population density data by combining information from two separate XML files. It assumes a specific XML structure where the `<data>` element contains `<record>` elements, and each `<record>` has fields with attributes specifying the type of data (e.g., "Country or Area" and "Value"). The script is flexible enough to handle different encodings or filenames by modifying the `areafn` and `popfn` variables.
